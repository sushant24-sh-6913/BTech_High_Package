```python
# Target: High Package Logic
name = input("Enter your name: ")
target_salary = float(input("Enter your target starting salary (in LPA): "))

print(f"\n--- MISSION START: {name.upper()} ---")
print(f"To hit {target_salary} LPA, we need to master 3 things:")
print("1. Data Structures (DSA)")
print("2. System Design")
print("3. AI/ML Specialization")

monthly_goal = target_salary / 12
print(f"\nYour target monthly income: ₹{monthly_goal:.2f} Lakhs")
```

```python
# The "Wealth Compounder" Loop
current_salary = 15  # Starting at 15 LPA (Our base target)
growth_rate = 1.50   # 50% increase every year for high-performers

print(f"Year 0: {current_salary} LPA")

for year in range(1, 6):
    current_salary = current_salary * growth_rate
    print(f"Year {year}: {current_salary:.2f} LPA")

print("\nConclusion: Skills + Consistency = Exponential Growth.")
```

```python
while True:
    m = float(input("Enter first number: "))
    op = input("Enter operator (+, -, *, /): ")
    n = float(input("Enter second number: "))

    if op == '+':
        print("Result:", m + n)
    elif op == '-':
        print("Result:", m - n)
    elif op == '*':
        print("Result:", m * n)
    elif op == '/':
        if n == 0:
            print("Error: Cannot divide by zero")
        else:
            print("Result:", m / n)
    else:
        print("Error: Invalid operator")

    again = input("Calculate again? (y/n): ")
    if again.lower() != 'y':
        break

print("Goodbye!")
```
